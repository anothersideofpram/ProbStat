```python
import pandas as pd
```

```python
dataraw = pd.read_excel('Pramana.xlsx', usecols='A:G')
print(dataraw)
```

```python
datafrq = pd.crosstab(index=dataraw['Grade'], columns='Frekuensi')
print(datafrq)
```

```python
datafrq.plot(title='Distribusi Grade - Pramana', marker='o')
```

```python
datafrq.plot(kind='bar', title='Distribusi Grade - Pramana', color='blue')
```

```python
datafrq.plot(kind='pie', y='Frekuensi', autopct='%1.1f%%', legend=False,
             title='Persentase Grade - Pramana')
```

```python
dataraw['Final Score'] = pd.to_numeric(dataraw['Final Score'], errors='coerce')
dt = dataraw['Final Score']

stats = dt.describe()
stats['Standard Error'] = dt.sem()
stats['Median'] = dt.median()
stats['Mode'] = dt.mode().iloc[0]
stats['Variance'] = dt.var()
stats['Range'] = dt.max() - dt.min()
stats['Skewness'] = dt.skew()
stats['Kurtosis'] = dt.kurtosis()
print(stats)
```
